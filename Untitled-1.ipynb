### 2(b). Phân loại các biến

**Biến định tính:**
- Mã sinh viên: biến định danh.
- Giới tính: biến định tính có hai nhóm Nam và Nữ.
- Ngành học: biến định tính phân loại theo ngành.
- Phương pháp học chính: biến định tính phân loại theo phương pháp học.

**Biến định lượng:**
- Tuổi: biến định lượng rời rạc nếu ghi nhận theo số năm nguyên.
- Thời gian tự học (giờ/ngày): biến định lượng liên tục.
- Thời gian ngủ (giờ/ngày): biến định lượng liên tục.
- GPA (thang 4): biến định lượng.

Trong các phần tiếp theo, GPA được sử dụng để ước lượng kỳ vọng và phương sai tổng thể; giới tính được sử dụng để ước lượng tỷ lệ sinh viên nữ và so sánh GPA giữa hai nhóm Nam và Nữ.

```python
# Mức ý nghĩa
alpha = 0.05

# Biến định lượng: GPA
gpa = df["GPA (thang 4)"].dropna().astype(float)

# Biến định tính: Giới tính
gender = df["Giới tính"].dropna()

print("Số quan sát GPA:", len(gpa))
print("Trung bình mẫu GPA:", round(gpa.mean(), 4))
print("Độ lệch chuẩn mẫu GPA:", round(gpa.std(ddof=1), 4))
print("Phương sai mẫu GPA:", round(gpa.var(ddof=1), 4))

print("\nPhân bố giới tính:")
print(gender.value_counts())
```

### 3(a). Ước lượng kỳ vọng tổng thể bằng khoảng tin cậy 95%

Chọn biến định lượng là GPA của sinh viên.

Do chưa biết phương sai tổng thể, sử dụng phân phối Student để xây dựng khoảng tin cậy cho kỳ vọng tổng thể.

**Công thức:**

\[
\bar{x}\pm t_{1-\alpha/2;n-1}\frac{s}{\sqrt n}
\]

Trong đó:
- \(\bar{x}\): trung bình mẫu.
- \(s\): độ lệch chuẩn mẫu.
- \(n\): số quan sát hợp lệ.
- \(t_{1-\alpha/2;n-1}\): giá trị tới hạn của phân phối Student với \(n-1\) bậc tự do.

**Điều kiện áp dụng:** các quan sát độc lập và mẫu được lấy theo phương pháp phù hợp. Tổng thể nên gần phân phối chuẩn hoặc cỡ mẫu đủ lớn để sử dụng xấp xỉ theo định lý giới hạn trung tâm.

```python
# Kích thước mẫu
n = len(gpa)

# Trung bình và độ lệch chuẩn mẫu
x_bar = gpa.mean()
s = gpa.std(ddof=1)

# Giá trị tới hạn Student
t_crit = stats.t.ppf(
    1 - alpha / 2,
    df=n - 1
)

# Sai số biên
margin = t_crit * s / np.sqrt(n)

# Khoảng tin cậy 95%
ci_mean = (
    x_bar - margin,
    x_bar + margin
)

print("CÂU 3(a): KHOẢNG TIN CẬY CHO KỲ VỌNG GPA")
print(f"Cỡ mẫu n = {n}")
print(f"Trung bình mẫu = {x_bar:.4f}")
print(f"Độ lệch chuẩn mẫu = {s:.4f}")
print(f"Giá trị t tới hạn = {t_crit:.4f}")
print(f"Sai số biên = {margin:.4f}")

print(
    f"Khoảng tin cậy 95%: "
    f"({ci_mean[0]:.4f}; {ci_mean[1]:.4f})"
)
```

### Nhận xét câu 3(a)

Từ mẫu gồm 100 sinh viên, ta tính được trung bình mẫu GPA và xây dựng khoảng tin cậy 95% cho kỳ vọng GPA tổng thể.

Với mức tin cậy 95%, khoảng tin cậy được tính từ dữ liệu là khoảng ước lượng cho GPA trung bình của tổng thể sinh viên giả định mà mẫu đại diện.

Khoảng tin cậy này ước lượng giá trị trung bình của tổng thể, không phải khoảng chứa 95% điểm GPA của từng sinh viên.

### 3(b). Ước lượng phương sai tổng thể bằng khoảng tin cậy 95%

Tiếp tục sử dụng biến GPA để ước lượng phương sai tổng thể.

**Công thức:**

\[
\left(
\frac{(n-1)s^2}{\chi^2_{1-\alpha/2;n-1}},
\frac{(n-1)s^2}{\chi^2_{\alpha/2;n-1}}
\right)
\]

Trong đó:
- \(s^2\): phương sai mẫu.
- \(n\): số quan sát.
- \(\chi^2_{\alpha/2;n-1}\) và \(\chi^2_{1-\alpha/2;n-1}\): các phân vị của phân phối Chi-bình phương.

**Điều kiện áp dụng:** mẫu ngẫu nhiên, các quan sát độc lập và tổng thể có phân phối chuẩn. Khoảng tin cậy cho phương sai đặc biệt nhạy với giả định phân phối chuẩn.

```python
# Cỡ mẫu và phương sai mẫu
n = len(gpa)
s2 = gpa.var(ddof=1)

# Bậc tự do
dfree = n - 1

# Hai phân vị Chi-bình phương
chi2_low = stats.chi2.ppf(
    alpha / 2,
    df=dfree
)

chi2_high = stats.chi2.ppf(
    1 - alpha / 2,
    df=dfree
)

# Khoảng tin cậy 95% cho phương sai
ci_var = (
    dfree * s2 / chi2_high,
    dfree * s2 / chi2_low
)

print("CÂU 3(b): KHOẢNG TIN CẬY CHO PHƯƠNG SAI GPA")
print(f"Cỡ mẫu n = {n}")
print(f"Phương sai mẫu s² = {s2:.4f}")
print(f"Bậc tự do = {dfree}")
print(f"Phân vị Chi-bình phương dưới = {chi2_low:.4f}")
print(f"Phân vị Chi-bình phương trên = {chi2_high:.4f}")

print(
    f"Khoảng tin cậy 95%: "
    f"({ci_var[0]:.4f}; {ci_var[1]:.4f})"
)
```

### Nhận xét câu 3(b)

Khoảng tin cậy 95% vừa tính được dùng để ước lượng phương sai GPA của tổng thể sinh viên giả định.

Phương sai thể hiện mức độ phân tán của điểm GPA quanh giá trị trung bình. Phương sai càng lớn thì điểm GPA trong tổng thể càng phân tán.

Kết quả cần được diễn giải với điều kiện tổng thể GPA có phân phối chuẩn. Vì bộ dữ liệu đang sử dụng là dữ liệu mô phỏng, kết quả chủ yếu minh họa phương pháp tính khoảng tin cậy bằng phân phối Chi-bình phương.

### 3(c). Ước lượng tỷ lệ tổng thể bằng khoảng tin cậy 95%

Chọn biến định tính nhị phân là giới tính. Quy ước nhóm cần ước lượng là sinh viên Nữ, nhóm còn lại là sinh viên Nam.

**Công thức:**

\[
\hat p\pm z_{1-\alpha/2}
\sqrt{\frac{\hat p(1-\hat p)}{n}}
\]

Trong đó:
- \(\hat p=x/n\): tỷ lệ mẫu sinh viên Nữ.
- \(x\): số sinh viên Nữ.
- \(n\): số sinh viên có dữ liệu giới tính hợp lệ.
- \(z_{1-\alpha/2}\): giá trị tới hạn của phân phối chuẩn tắc, bằng khoảng 1,96 khi mức tin cậy là 95%.

**Điều kiện áp dụng:** các quan sát độc lập và số lượng kỳ vọng của cả hai nhóm đủ lớn để sử dụng xấp xỉ chuẩn. Nếu điều kiện không thỏa mãn, có thể sử dụng khoảng Wilson hoặc khoảng nhị thức chính xác.

```python
# Lấy dữ liệu giới tính hợp lệ
gender = df["Giới tính"].dropna()

# Cỡ mẫu
n_gender = len(gender)

# Số sinh viên Nữ
x_female = (gender == "Nữ").sum()

# Tỷ lệ mẫu
p_hat = x_female / n_gender

# Giá trị tới hạn chuẩn tắc
z_crit = stats.norm.ppf(1 - alpha / 2)

# Sai số biên
se_p = np.sqrt(
    p_hat * (1 - p_hat) / n_gender
)

margin_p = z_crit * se_p

# Khoảng tin cậy 95%
ci_prop = (
    max(0, p_hat - margin_p),
    min(1, p_hat + margin_p)
)

print("CÂU 3(c): KHOẢNG TIN CẬY CHO TỶ LỆ SINH VIÊN NỮ")
print(f"Cỡ mẫu = {n_gender}")
print(f"Số sinh viên Nữ = {x_female}")
print(f"Tỷ lệ mẫu = {p_hat:.4f}")
print(f"Tỷ lệ mẫu theo phần trăm = {p_hat * 100:.2f}%")

print(
    f"Khoảng tin cậy 95%: "
    f"({ci_prop[0]:.4f}; {ci_prop[1]:.4f})"
)

print(
    f"Khoảng theo phần trăm: "
    f"({ci_prop[0] * 100:.2f}%; "
    f"{ci_prop[1] * 100:.2f}%)"
)

print("n*p =", round(n_gender * p_hat, 2))
print("n*(1-p) =", round(n_gender * (1-p_hat), 2))
```

### Nhận xét câu 3(c)

Từ số lượng sinh viên Nữ trong mẫu, ta tính được tỷ lệ mẫu và xây dựng khoảng tin cậy 95% cho tỷ lệ sinh viên Nữ trong tổng thể giả định.

Khoảng tin cậy được biểu diễn theo tỷ lệ hoặc phần trăm. Khi diễn giải, cần lưu ý đây là khoảng ước lượng tỷ lệ của tổng thể, không phải khoảng chứa 95% số sinh viên trong mẫu.

Kết quả phụ thuộc vào tính đại diện của mẫu và điều kiện áp dụng phương pháp xấp xỉ chuẩn.

### 3(d). So sánh kỳ vọng GPA giữa hai nhóm

Chọn biến định tính giới tính để chia sinh viên thành hai nhóm:

- Nhóm 1: sinh viên Nam.
- Nhóm 2: sinh viên Nữ.

Biến định lượng được so sánh là GPA theo thang điểm 4. Mục tiêu là xây dựng khoảng tin cậy 95% cho hiệu hai kỳ vọng tổng thể:

\[
\mu_1-\mu_2=\mu_{\text{Nam}}-\mu_{\text{Nữ}}
\]

Sử dụng khoảng tin cậy Welch vì phương pháp này không yêu cầu phương sai của hai nhóm bằng nhau.

**Công thức:**

\[
(\bar x_1-\bar x_2)
\pm t_{1-\alpha/2;\nu}
\sqrt{\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}}
\]

Bậc tự do xấp xỉ:

\[
\nu=
\frac{\left(\frac{s_1^2}{n_1}+
\frac{s_2^2}{n_2}\right)^2}
{\frac{(s_1^2/n_1)^2}{n_1-1}
+\frac{(s_2^2/n_2)^2}{n_2-1}}
\]

Trong đó \(n_1,n_2\) là cỡ mẫu; \(\bar x_1,\bar x_2\) là trung bình mẫu; \(s_1^2,s_2^2\) là phương sai mẫu của hai nhóm.

**Điều kiện áp dụng:** hai nhóm độc lập, các quan sát độc lập trong từng nhóm và dữ liệu trong mỗi nhóm gần phân phối chuẩn hoặc có cỡ mẫu đủ lớn.

```python
# Tách GPA theo giới tính
group1 = df.loc[
    df["Giới tính"] == "Nam",
    "GPA (thang 4)"
].dropna().astype(float)

group2 = df.loc[
    df["Giới tính"] == "Nữ",
    "GPA (thang 4)"
].dropna().astype(float)

# Cỡ mẫu mỗi nhóm
n1 = len(group1)
n2 = len(group2)

# Trung bình mẫu
mean1 = group1.mean()
mean2 = group2.mean()

# Phương sai mẫu
var1 = group1.var(ddof=1)
var2 = group2.var(ddof=1)

# Hiệu trung bình mẫu
difference = mean1 - mean2

# Sai số chuẩn Welch
se_diff = np.sqrt(
    var1 / n1 + var2 / n2
)

# Bậc tự do Welch-Satterthwaite
df_welch = (
    (var1 / n1 + var2 / n2) ** 2
    /
    (
        (var1 / n1) ** 2 / (n1 - 1)
        +
        (var2 / n2) ** 2 / (n2 - 1)
    )
)

# Giá trị tới hạn Student
t_crit_diff = stats.t.ppf(
    1 - alpha / 2,
    df=df_welch
)

# Sai số biên
margin_diff = t_crit_diff * se_diff

# Khoảng tin cậy
ci_diff = (
    difference - margin_diff,
    difference + margin_diff
)

print("CÂU 3(d): SO SÁNH KỲ VỌNG GPA GIỮA NAM VÀ NỮ")
print("\nNhóm Nam:")
print(f"n1 = {n1}")
print(f"Trung bình GPA = {mean1:.4f}")
print(f"Phương sai mẫu = {var1:.4f}")

print("\nNhóm Nữ:")
print(f"n2 = {n2}")
print(f"Trung bình GPA = {mean2:.4f}")
print(f"Phương sai mẫu = {var2:.4f}")

print(f"\nHiệu trung bình (Nam - Nữ) = {difference:.4f}")
print(f"Sai số chuẩn = {se_diff:.4f}")
print(f"Bậc tự do Welch = {df_welch:.2f}")

print(
    f"Khoảng tin cậy 95% cho μ1 - μ2: "
    f"({ci_diff[0]:.4f}; {ci_diff[1]:.4f})"
)
```

```python
if ci_diff[0] > 0:
    print(
        "Kết luận: Khoảng tin cậy nằm hoàn toàn trên 0. "
        "GPA trung bình của nhóm Nam được ước lượng cao hơn nhóm Nữ."
    )

elif ci_diff[1] < 0:
    print(
        "Kết luận: Khoảng tin cậy nằm hoàn toàn dưới 0. "
        "GPA trung bình của nhóm Nam được ước lượng thấp hơn nhóm Nữ."
    )

else:
    print(
        "Kết luận: Khoảng tin cậy chứa 0. "
        "Chưa đủ bằng chứng ở mức ý nghĩa 5% "
        "để kết luận kỳ vọng GPA của hai nhóm khác nhau."
    )
```

### Nhận xét câu 3(d)

Khoảng tin cậy 95% của hiệu hai kỳ vọng GPA được sử dụng để đánh giá sự khác biệt giữa nhóm sinh viên Nam và Nữ.

Nếu khoảng tin cậy nằm hoàn toàn phía trên 0, kết quả gợi ý GPA trung bình của nhóm Nam cao hơn nhóm Nữ. Nếu khoảng nằm hoàn toàn phía dưới 0, kết quả gợi ý GPA trung bình của nhóm Nam thấp hơn nhóm Nữ. Nếu khoảng chứa 0, chưa có đủ bằng chứng thống kê ở mức ý nghĩa 5% để kết luận hai kỳ vọng khác nhau.

Kết quả chỉ phản ánh dữ liệu mô phỏng đang xét và không đủ cơ sở để kết luận về tất cả sinh viên trong thực tế.

```python
summary = pd.DataFrame([
    {
        "Nội dung": "Kỳ vọng GPA",
        "Ước lượng": x_bar,
        "Cận dưới 95%": ci_mean[0],
        "Cận trên 95%": ci_mean[1]
    },
    {
        "Nội dung": "Phương sai GPA",
        "Ước lượng": s2,
        "Cận dưới 95%": ci_var[0],
        "Cận trên 95%": ci_var[1]
    },
    {
        "Nội dung": "Tỷ lệ sinh viên Nữ",
        "Ước lượng": p_hat,
        "Cận dưới 95%": ci_prop[0],
        "Cận trên 95%": ci_prop[1]
    },
    {
        "Nội dung": "Hiệu GPA (Nam - Nữ)",
        "Ước lượng": difference,
        "Cận dưới 95%": ci_diff[0],
        "Cận trên 95%": ci_diff[1]
    }
])

display(summary.round(4))
```

### 4(b). Điều kiện áp dụng

Khoảng tin cậy cho kỳ vọng sử dụng phân phối Student khi phương sai tổng thể chưa biết; cần các quan sát độc lập và giả định phân phối chuẩn hoặc cỡ mẫu đủ lớn.

Khoảng tin cậy cho phương sai sử dụng phân phối Chi-bình phương và yêu cầu tổng thể có phân phối chuẩn.

Khoảng tin cậy cho tỷ lệ sử dụng xấp xỉ chuẩn khi số lượng quan sát kỳ vọng trong cả hai nhóm đủ lớn.

Khoảng tin cậy cho hiệu hai kỳ vọng sử dụng phương pháp Welch, phù hợp khi hai nhóm độc lập và không cần giả định phương sai hai nhóm bằng nhau.

### 4(c). Kết luận

Bài tập đã sử dụng Python để phân tích bộ dữ liệu mô phỏng gồm 100 sinh viên. Thông qua các thư viện Pandas, NumPy và SciPy, bài làm đã thực hiện ước lượng khoảng tin cậy 95% cho kỳ vọng GPA, phương sai GPA, tỷ lệ sinh viên Nữ và hiệu kỳ vọng GPA giữa hai nhóm Nam và Nữ.

Các kết quả được trình bày bằng giá trị ước lượng, cận dưới, cận trên và phần diễn giải theo bối cảnh dữ liệu. Việc sử dụng khoảng tin cậy giúp thể hiện mức độ không chắc chắn khi ước lượng các tham số tổng thể từ mẫu.

Do dữ liệu được mô phỏng, kết quả có ý nghĩa minh họa cho phương pháp thống kê suy luận, không đại diện cho kết quả khảo sát sinh viên ngoài thực tế.
